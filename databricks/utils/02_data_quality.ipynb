```python
from pyspark.sql import DataFrame
from pyspark.sql import functions as F
```

```python
def check_required_fields(
    df: DataFrame,
    columns: list[str]
) -> None:
    for column in columns:
        null_count = df.filter(F.col(column).isNull()).count()

        assert null_count == 0, (
            f"Quality check failed: {column} contains "
            f"{null_count} null values."
        )

    print(f"Required fields validated: {columns}")
```

```python
def check_unique_keys(
    df: DataFrame,
    columns: list[str]
) -> None:
    duplicate_count = (
        df.groupBy(*columns)
        .count()
        .filter(F.col("count") > 1)
        .count()
    )

    assert duplicate_count == 0, (
        f"Quality check failed: {duplicate_count} duplicate keys "
        f"found for {columns}."
    )

    print(f"Unique keys validated: {columns}")
```

```python
def check_row_count(
    df: DataFrame,
    expected_count: int
) -> None:
    actual_count = df.count()

    assert actual_count == expected_count, (
        f"Quality check failed: expected {expected_count} rows, "
        f"found {actual_count}."
    )

    print(f"Row count validated: {actual_count}")
```
